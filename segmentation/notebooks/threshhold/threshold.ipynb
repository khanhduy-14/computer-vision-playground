```python
import os
import numpy as np
import matplotlib.pyplot as plt

from PIL import Image

print("CWD: " + os.getcwd())
img = Image.open("../../assets/threshold/cat.jpg").convert('RGB')

plt.imshow(img)
plt.axis('off')
plt.show()


# plt.plot(numbers)
# plt.show()
```

```python
img_array = np.array(img)

R = img_array[:, :, 0]
G = img_array[:, :, 1]
B = img_array[:, :, 2]


gray = 0.299 * R + 0.587 * G + 0.114 * B

gray = gray.astype(np.uint8)



plt.imshow(gray, cmap='gray')
plt.axis('off')
plt.show()
```

```python
threshold = 200
gray[gray>threshold] = 255
gray[gray<=threshold] = 0


fig, axes = plt.subplots(1,2 ,figsize=(10,5))

axes[0].imshow(img)
axes[0].set_title('Original cat image')

axes[1].imshow(gray, cmap='gray')
axes[1].set_title('Segmented image by threshold based')

plt.show()
```
